# The Great Job Interview: A MadLib

## How to Play

1. Run the **Gather Your Words** cell and answer five prompts.
2. Run the **The Story** cell to hear what happens.
3. Want a new story? Run both cells again with different words.

## Gather Your Words

```python
print("Let's write a story! Give me five words.")

animal = input("Give me an animal: ")
color = input("Give me a color: ")
city = input("Give me a city in Washington: ")
food = input("Give me a food: ")
adjective = input("Give me an adjective: ")
```

## The Story

```python
from IPython.display import Markdown, display

story = f"""
### The Great Job Interview

One rainy morning, a **{adjective}** **{animal}** put on a tiny **{color}** tie and set off for **{city}**
to interview for its dream job.

On the way, it got so nervous that it stopped for a snack and ate an entire **{food}**.
That turned out to be a bad idea, because the {food} was *very* messy.

When the **{animal}** walked into the interview, the manager looked up and said,
"You have {food} on your tie."

The **{animal}** smiled, stood tall, and replied, "That's just my secret to success."

**They hired it on the spot.**
"""

display(Markdown(story))
```
